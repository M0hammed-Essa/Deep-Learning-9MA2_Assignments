# Task 1: Install PyTorch 

```python
import torch
import torchvision
import torchvision.transforms as transforms

print("PyTorch version:", torch.__version__)
print("Torchvision version:", torchvision.__version__)
```

# Task 2: Inspect the Dataset

```python

# Convert images to tensors and normalize them
transform = transforms.Compose([
    transforms.ToTensor()
])

# Download and load the training dataset
train_dataset = torchvision.datasets.MNIST(
    root="./data",
    train=True,
    download=True,
    transform=transform
)

# Download and load the testing dataset
test_dataset = torchvision.datasets.MNIST(
    root="./data",
    train=False,
    download=True,
    transform=transform
)

print("Training dataset size:", len(train_dataset))
print("Testing dataset size:", len(test_dataset))
```

# Task 3: Inspect the Dataset

```python
import matplotlib.pyplot as plt

plt.figure(figsize=(12, 5))

for i in range(10):
    image, label = train_dataset[i]

    plt.subplot(2, 5, i + 1)
    plt.imshow(image.squeeze(), cmap="gray")
    plt.title(f"Label: {label}")
    plt.axis("off")

plt.tight_layout()
plt.show()
```

```python
from torch.utils.data import DataLoader

batch_size = 32

train_loader = DataLoader(
    train_dataset,
    batch_size=batch_size,
    shuffle=True
)

test_loader = DataLoader(
    test_dataset,
    batch_size=batch_size,
    shuffle=False
)

images, labels = next(iter(train_loader))

print("Batch image shape:", images.shape)
print("Batch labels shape:", labels.shape)
```

# Task 4: Creating a Simple Neural Network

```python
import torch.nn as nn

class SimpleNN(nn.Module):
    def __init__(self):
        super(SimpleNN, self).__init__()

        # Input: 28 x 28 = 784
        self.fc1 = nn.Linear(784, 128)

        # Output: 10 classes (digits 0-9)
        self.fc2 = nn.Linear(128, 10)

    def forward(self, x):

        # Flatten:
        # [batch, 1, 28, 28] -> [batch, 784]
        x = torch.flatten(x, start_dim=1)

        # Hidden layer + ReLU
        x = torch.relu(self.fc1(x))

        # Output layer
        x = self.fc2(x)

        return x


# Create the network
model = SimpleNN()

print(model)
```

```python
images, labels = next(iter(train_loader))

output = model(images)

print("Input shape:", images.shape)
print("Output shape:", output.shape)
```

# Task 5: Training Function

```python
import torch.optim as optim

def train(model, train_loader, epochs=5, learning_rate=0.01, momentum=0.9):

    # Loss function
    criterion = nn.CrossEntropyLoss()

    # SGD optimizer
    optimizer = optim.SGD(
        model.parameters(),
        lr=learning_rate,
        momentum=momentum
    )

    loss_history = []
    accuracy_history = []

    for epoch in range(epochs):

        model.train()

        total_loss = 0
        correct = 0
        total = 0

        for batch_idx, (images, labels) in enumerate(train_loader):

            # Clear previous gradients
            optimizer.zero_grad()

            # Forward pass
            outputs = model(images)

            # Calculate loss
            loss = criterion(outputs, labels)

            # Backpropagation
            loss.backward()

            # Update weights
            optimizer.step()

            # Statistics
            total_loss += loss.item()

            _, predicted = torch.max(outputs, 1)

            total += labels.size(0)
            correct += (predicted == labels).sum().item()

            # Debug information
            if batch_idx % 800 == 0:
                accuracy = 100 * correct / total

                print(
                    f"Epoch [{epoch+1}/{epochs}] "
                    f"Batch [{batch_idx}/{len(train_loader)}] "
                    f"Loss: {loss.item():.4f} "
                    f"Accuracy: {accuracy:.2f}%"
                )

        # Average loss and accuracy for epoch
        epoch_loss = total_loss / len(train_loader)
        epoch_accuracy = 100 * correct / total

        loss_history.append(epoch_loss)
        accuracy_history.append(epoch_accuracy)

        print(
            f"Epoch {epoch+1} completed - "
            f"Loss: {epoch_loss:.4f}, "
            f"Accuracy: {epoch_accuracy:.2f}%"
        )

    return loss_history, accuracy_history
```

# Task 6: Train the Network

```python
torch.manual_seed(42)

# Create a new model
model = SimpleNN()

# Train the model
loss_history, accuracy_history = train(
    model,
    train_loader,
    epochs=5,
    learning_rate=0.01,
    momentum=0.9
)
```

```python
plt.figure(figsize=(8, 5))

plt.plot(
    range(1, len(loss_history) + 1),
    loss_history,
    marker="o"
)

plt.xlabel("Epoch")
plt.ylabel("Loss")
plt.title("Training Loss")
plt.grid()
plt.show()
```

```python
plt.figure(figsize=(8, 5))

plt.plot(
    range(1, len(accuracy_history) + 1),
    accuracy_history,
    marker="o"
)

plt.xlabel("Epoch")
plt.ylabel("Accuracy (%)")
plt.title("Training Accuracy")
plt.grid()
plt.show()
```

# Task 7: Testing the Network

```python
def test(model, test_loader):

    model.eval()

    total_loss = 0
    correct = 0
    total = 0

    with torch.no_grad():

        for images, labels in test_loader:

            outputs = model(images)

            _, predicted = torch.max(outputs, 1)

            total += labels.size(0)
            correct += (predicted == labels).sum().item()

    accuracy = 100 * correct / total

    print(f"Test Accuracy: {accuracy:.2f}%")

    return accuracy


test_accuracy = test(model, test_loader)
```

```python
# Get one batch from test data
images, labels = next(iter(test_loader))

# Make predictions
model.eval()

with torch.no_grad():
    outputs = model(images)
    _, predictions = torch.max(outputs, 1)

# Display 10 images
plt.figure(figsize=(12, 5))

for i in range(10):

    plt.subplot(2, 5, i + 1)

    plt.imshow(images[i].squeeze(), cmap="gray")

    plt.title(
        f"Actual: {labels[i].item()}\n"
        f"Predicted: {predictions[i].item()}"
    )

    plt.axis("off")

plt.tight_layout()
plt.show()
```

# Task 8: Tuning the Parameters and Model Architecture

## 8.0 A flexible network so we can change depth/width easily

Instead of hardcoding one architecture, `FlexibleNN` takes a list of hidden-layer sizes, so `[128]` gives the same network as Task 4, `[256, 128]` gives two hidden layers, etc.

```python
class FlexibleNN(nn.Module):
    def __init__(self, hidden_layers=[128]):
        """
        hidden_layers: list of ints.
        e.g. [128]        -> one hidden layer of 128 neurons (same as Task 4)
             [256, 128]   -> two hidden layers
             [256,128,64] -> three hidden layers
        """
        super(FlexibleNN, self).__init__()
        layers = []
        input_size = 784
        for h in hidden_layers:
            layers.append(nn.Linear(input_size, h))
            layers.append(nn.ReLU())
            input_size = h
        layers.append(nn.Linear(input_size, 10))  # output layer, 10 classes
        self.net = nn.Sequential(*layers)

    def forward(self, x):
        x = torch.flatten(x, start_dim=1)
        return self.net(x)


def evaluate_quiet(model, loader):
    """Same as test(), but no printing - used for scoring many runs during search."""
    model.eval()
    criterion = nn.CrossEntropyLoss()
    correct, total, total_loss = 0, 0, 0
    with torch.no_grad():
        for images, labels in loader:
            outputs = model(images)
            loss = criterion(outputs, labels)
            total_loss += loss.item()
            _, predicted = torch.max(outputs, 1)
            total += labels.size(0)
            correct += (predicted == labels).sum().item()
    return total_loss / len(loader), 100 * correct / total


def train_quiet(model, loader, epochs, learning_rate, momentum):
    """Same training loop as train(), but no printing - used for the grid search."""
    criterion = nn.CrossEntropyLoss()
    optimizer = optim.SGD(model.parameters(), lr=learning_rate, momentum=momentum)
    for epoch in range(epochs):
        model.train()
        for images, labels in loader:
            optimizer.zero_grad()
            outputs = model(images)
            loss = criterion(outputs, labels)
            loss.backward()
            optimizer.step()
    return model
```

## 8.1 Tuning learning rate and momentum

Fix the architecture at one hidden layer of 128 neurons (same as Task 4) and sweep over learning rate and momentum. Each combination is trained for a few epochs (enough to separate good settings from bad ones) and scored on the test set.

```python
import itertools
import pandas as pd

learning_rates = [0.001, 0.01, 0.05, 0.1]
momentums = [0.0, 0.5, 0.9]
SEARCH_EPOCHS = 3  # short runs just to rank the combinations

lr_momentum_results = []

for lr, mom in itertools.product(learning_rates, momentums):
    torch.manual_seed(42)
    model = FlexibleNN(hidden_layers=[128])
    model = train_quiet(model, train_loader, epochs=SEARCH_EPOCHS,
                         learning_rate=lr, momentum=mom)
    test_loss, test_acc = evaluate_quiet(model, test_loader)

    lr_momentum_results.append({
        "lr": lr, "momentum": mom,
        "test_loss": test_loss, "test_accuracy": test_acc
    })
    print(f"lr={lr:<6} momentum={mom:<4} -> "
          f"test_loss={test_loss:.4f}, test_accuracy={test_acc:.2f}%")

lr_momentum_df = pd.DataFrame(lr_momentum_results).sort_values(
    by=["test_accuracy", "test_loss"], ascending=[False, True]
).reset_index(drop=True)

print("\nRanked results (best first):")
lr_momentum_df
```

## 8.2 Tuning the architecture (number of layers and neurons)

Take the best learning rate / momentum pair from 8.1 and hold it fixed while sweeping over different numbers of hidden layers and neurons per layer.

```python
best_lr_row = lr_momentum_df.iloc[0]
best_lr = float(best_lr_row["lr"])
best_momentum = float(best_lr_row["momentum"])
print(f"Best learning rate/momentum from 8.1: lr={best_lr}, momentum={best_momentum}")

architectures = {
    "1x64":         [64],
    "1x128":        [128],
    "1x256":        [256],
    "2x128-64":     [128, 64],
    "2x256-128":    [256, 128],
    "3x256-128-64": [256, 128, 64],
}

arch_results = []

for name, layers_cfg in architectures.items():
    torch.manual_seed(42)
    model = FlexibleNN(hidden_layers=layers_cfg)
    model = train_quiet(model, train_loader, epochs=SEARCH_EPOCHS,
                         learning_rate=best_lr, momentum=best_momentum)
    test_loss, test_acc = evaluate_quiet(model, test_loader)

    arch_results.append({
        "architecture": name, "layers": layers_cfg,
        "test_loss": test_loss, "test_accuracy": test_acc
    })
    print(f"{name:<15} -> test_loss={test_loss:.4f}, test_accuracy={test_acc:.2f}%")

arch_df = pd.DataFrame(arch_results).sort_values(
    by=["test_accuracy", "test_loss"], ascending=[False, True]
).reset_index(drop=True)

print("\nRanked results (best first):")
arch_df
```

## 8.3 Final best model

Combine the best learning rate/momentum (8.1) with the best architecture (8.2), then train it properly (more epochs, using the same `train()` function and debug printing from Task 5/6) for the final reported result.

```python
best_arch_row = arch_df.iloc[0]
best_layers = best_arch_row["layers"]

print("=== Best configuration found by the search ===")
print(f"Architecture : {best_arch_row['architecture']}  {best_layers}")
print(f"Learning rate: {best_lr}")
print(f"Momentum     : {best_momentum}")

FINAL_EPOCHS = 15

torch.manual_seed(42)
best_model = FlexibleNN(hidden_layers=best_layers)

best_loss_history, best_accuracy_history = train(
    best_model,
    train_loader,
    epochs=FINAL_EPOCHS,
    learning_rate=best_lr,
    momentum=best_momentum
)

final_test_accuracy = test(best_model, test_loader)

print("\n=== FINAL BEST RESULT ===")
print(f"Architecture      : {best_layers}")
print(f"Learning rate     : {best_lr}")
print(f"Momentum          : {best_momentum}")
print(f"Final train loss  : {best_loss_history[-1]:.4f}")
print(f"Final train acc.  : {best_accuracy_history[-1]:.2f}%")
print(f"Final test acc.   : {final_test_accuracy:.2f}%")
```

```python
plt.figure(figsize=(8, 5))
plt.plot(range(1, len(best_loss_history) + 1), best_loss_history, marker="o")
plt.xlabel("Epoch")
plt.ylabel("Loss")
plt.title(f"Best Model Training Loss (layers={best_layers}, lr={best_lr}, momentum={best_momentum})")
plt.grid()
plt.show()

plt.figure(figsize=(8, 5))
plt.plot(range(1, len(best_accuracy_history) + 1), best_accuracy_history, marker="o", color="green")
plt.xlabel("Epoch")
plt.ylabel("Accuracy (%)")
plt.title(f"Best Model Training Accuracy (layers={best_layers}, lr={best_lr}, momentum={best_momentum})")
plt.grid()
plt.show()
```
